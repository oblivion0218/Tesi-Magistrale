### Definizione di Reattività

La reattività $\rho$ è definita in relazione al fattore di moltiplicazione efficace $k_{eff}$:

$$
\rho = \text{Reattività} = \frac{k_{eff} - 1}{k_{eff}}
$$


Invertendo poi la relazione per esplicitare $k_{eff}$:

$$
k_{eff} = \frac{1}{1 - \rho}
$$


### Scomposizione della Reattività

La reattività totale $\rho_{TOT}$ è data dalla somma della reattività iniziale $\rho_0$ e delle variazioni $\Delta \rho$:

$$
\rho_{TOT} = \rho_0 + \Delta \rho
$$

Dove $\rho_0$ è definita rispetto allo stato iniziale:

$$
\rho_0 = \frac{k_{eff_0} - 1}{k_{eff_0}}
$$

mentre il $\Delta \rho $ viene dalla sommatoria dei coefficienti di feedback $\alpha_i$ per le variazioni di stato $\Delta x_i$

Sostituendo l'espressione della reattività totale (inclusa la sommatoria) nella formula di $k_{eff}$:

$$
k_{eff} = \frac{1}{1 - \left( \rho_0 + \sum \alpha_i \Delta x_i \right)}
$$

### 1. Effetto Doppler (Combustibile)
Il meccanismo dominante per la sicurezza intrinseca legato alla temperatura del combustibile $T_f$.

* **Fisica:** L'aumento dell'agitazione termica dei nuclei bersaglio allarga le risonanze di sezione d'urto (Doppler Broadening).
* **Dipendenza Spettrale:**
    * **Spettro Veloce (Fast):** Fondamentale nella regione delle risonanze non risolte (keV - MeV). L'allargamento aumenta la cattura parassita negli isotopi fertili ($^{238}U$, $^{232}Th$) e negli attinidi minori.
    * **Spettro Termico:** Agisce sulle risonanze risolte a bassa energia (eV). Aumenta la cattura, riducendo la probabilità di fuga dalla risonanza $p$.

$$
\alpha_{Doppler} = \frac{1}{k} \frac{dk}{dT_f} \approx \frac{A}{T_f^\gamma}
$$

Dove $\gamma$ dipende dallo spettro: $\gamma \approx 1$ per spettri veloci (ossidi misti/nitruri), $\gamma \approx 0.5$ per termici ($1/\sqrt{T}$).

### 2. Coefficiente di Densità del Refrigerante (Coolant Density/Void)
Risponde alla variazione di densità del fluido termovettore ($\rho_c$). La fisica cambia radicalmente in base allo spettro neutronico.

$$
\alpha_{\rho} = \frac{\partial \rho_{reaz}}{\partial \rho_c}
$$

#### A. Regime a Spettro Veloce (Tipico ADS a Pb/LBE)
La riduzione di densità del refrigerante provoca due effetti competitivi:
1.  **Indurimento Spettrale (Spectral Hardening):** Meno urti elastici/inelastici con il refrigerante $\rightarrow$ lo spettro medio si sposta a energie più alte.
    * *Effetto su $k$:* **Positivo**. Aumenta la fissione negli isotopi a soglia (Attinidi Minori pari).
2.  **Aumento delle Fughe (Leakage):** Il cammino libero medio $\lambda_{tr}$ aumenta.
    * *Effetto su $k$:* **Negativo**. I neutroni fuggono dal nocciolo.

La condizione di stabilità richiede che il *Leakage* domini sull'*Hardening*

#### B. Regime a Spettro Termico
La riduzione di densità riduce la moderazione.
* **Effetto:** Diminuisce il potere rallentante macroscopico $\xi \Sigma_s$. Lo spettro diventa meno termico, riducendo il fattore di utilizzo termico $f$ e la probabilità di fuga alla risonanza $p$.
* *Effetto su $k$:* Fortemente **Negativo** (sottomoderazione).

$$
k_{eff} \approx \eta \epsilon p f \implies \frac{\partial k}{k} \approx \frac{\partial p}{p} + \frac{\partial f}{f}
$$

### 3. Espansione Strutturale e del Combustibile
Negli ADS, spesso a geometrie compatte e spettri veloci, le variazioni geometriche impattano direttamente il Buckling geometrico.

* **Espansione Assiale (Fuel):** Il combustibile si allunga, riducendo la sua densità atomica effettiva nel volume attivo.
* **Espansione Radiale (Grid Plate):** La griglia di supporto si dilata allontanando gli elementi (aumento del *pitch*).

L'espansione aumenta le dimensioni ($R \to R(1+\frac{\Delta L}{L})$), ma riduce drasticamente la densità media macroscopica del nocciolo, aumentando il cammino libero medio di trasporto e favorendo la fuga dei neutroni (riduzione di $k_{eff}$).

### 4. Dinamica della Sorgente (Non è un feedback di reattività, ma di potenza)
Specifico per ADS sottocritici. La potenza termica $P$ dipende istantaneamente dalla reattività $\rho$ (che include i feedback sopra citati) e dalla sorgente $S$.

Se i feedback termici ($\alpha_{Doppler}$, $\alpha_{\rho}$) sono negativi, un aumento di temperatura rende $\rho(t)$ più negativo (es. da $-0.03$ a $-0.05$).

**Risultato:** A parità di sorgente $S$, la potenza $P$ diminuisce. Il sistema è autolimitante.
